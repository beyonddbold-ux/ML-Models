In the NLTK library, wordpunct_tokenize is a function that splits a text string into a list of alphabetic and non-alphabetic characters using a simple regular expression.

“It uses the regular expression \w+|[^\w\s]+ to split the input.” This means that all punctuation marks and continuous sequences of letters/numbers are cleanly isolated into individual tokens, regardless of their semantic meaning

Regular Expression Breakdown (\w+|[^\w\s]+)

\w+: Matches one or more word characters (letters, digits, and the underscore _). These form word tokens.

|: Acts as an "OR" operator, meaning the regex matches either a word or a punctuation block.

[^\w\s]+: Matches one or more characters that are not word characters (\w) and not whitespace (\s). This isolates any punctuation or symbols into their own separate tokens.

Instead of keeping punctuation attached to words, it separates all punctuation into its own tokens.

```python
# treebank_wt = nltk.TreebankWordTokenizer()
#splits words with standard contractions #examples   ‘don’t’  to   ‘do’ and ‘n’t’
treebank_wt = nltk.TreebankWordTokenizer()
words = treebank_wt.tokenize(s1)
print(words)
```

The class nltk.TreebankWordTokenizer() is an advanced regular-expression-based word tokenizer in the Natural Language Toolkit (NLTK) package that splits sentences into individual words and punctuation tokens based on the conventions of the Penn Treebank dataset. It is widely used in NLP because it processes standard English contractions, grammatical structures, and punctuation symbols much more intelligently than a simple whitespace split.
The commonly used nltk.word_tokenize() function is actually a wrapper that initializes and runs this specific tokenizer class behind the scenes.Core Tokenization RulesThe TreebankWordTokenizer follows distinct linguistic rules to break text down into tokens:
Splits contractions: It isolates clitics and shortened verb formats (e.g., "don't" transforms into ['do', "n't"], and "they'll" transforms into ['they', "'ll"]).
Isolates punctuation: Most punctuation symbols, including currency markers like $, are detached into their own unique tokens.
Conditional commas and quotes: Commas and single quotes are split out only when they are directly followed by whitespace characters.
Trailing periods: Periods found exactly at the conclusion of a line or sentence boundary are separated completely.

```python
s1 = "Good muffins cost $3.88\nin New York.  Please buy me ... two of them.\n\n Thanks."
words = treebank_wt.tokenize(s1)
print(words)
```

The Treebank tokenizer is a standard text segmentation method based on the Penn Treebank project. It splits standard English contractions, separates punctuation marks, and isolates end-of-line periods
