STEMMING

````python
from nltk.stem import PorterStemmer

# Initialize the Porter Stemmer
stemmer = PorterStemmer()

# Example words to demonstrate stemming
words = [
    "running", "runner", "runs",
    "beautiful", "beauty",
    "connecting", "connection", "connected",
    "generously", "general",
    "history", "historical", "historic"
]

print("Demonstrating Porter Stemmer:\n")
for word in words:
    stemmed_word = stemmer.stem(word)
    print(f"Original: {word.ljust(15)} Stemmed: {stemmed_word}")
````

**SNOWBALL STEMMER **                 
The Snowball stemmer (also known as the Porter2 stemmer) is a widely used algorithmic language processing tool that reduces words to their base or root form by stripping common suffixes.

**Key Characteristics**

Improved Framework: Developed by Martin Porter as an upgrade to his original 1980 Porter Stemmer.

Algorithmic Base: Uses a string-handling language called Snowball to define stemming rules.

Speed & Efficiency: Executes quickly because it relies on predefined linguistic rules rather than heavy database lookups.

Multi-Language Support: Works across many languages, including English, Spanish, French, German, and Russian.

````python
from nltk.stem import SnowballStemmer

# Initialize the Snowball Stemmer for English
snowball_stemmer = SnowballStemmer("english")

# Example words to demonstrate stemming
words_snowball = [
    "generously", "generation", "generalize", "generosity",
    "universities", "university",
    "corporations", "corporate",
    "agreement", "agreed",
    "studies", "studying", "student"
]

print("Demonstrating Snowball Stemmer (English):\n")
for word in words_snowball:
    stemmed_word = snowball_stemmer.stem(word)
    print(f"Original: {word.ljust(15)} Stemmed: {stemmed_word}")
````

Data Cleaning - Remove punctuations, remove numbers and Remove extra whitespace and convert to lowercase

````python
# regular expressions - This module provides regular expression matching operations
import re
# module string - A collection of string constants
import string

def clean_text(text):
    # Remove punctuation using str.maketrans(*args,**kwargs)
    # In Python, *args and **kwargs are special symbols used in function definitions to pass
    # a flexible, variable number of arguments: *args collects extra positional arguments into a tuple,
    # and **kwargs collects extra keyword (named) arguments into a dictionary
    text = text.translate(str.maketrans('', '', string.punctuation))
     # The str.maketrans(x, y, z) method creates a translation table for characters.
     # The first two arguments x and y are empty strings '', meaning no characters are being replaced with other characters.
     # The third argument z specifies characters to be deleted. Passing string.punctuation here maps every punctuation mark to None and delets all punctuation marks

    # Remove numbers
    text = re.sub(r'\d+', '', text)

    # Remove extra whitespace and convert to lowercase
    text = re.sub(r'\s+', ' ', text).strip().lower()
    return text

# Example usage
sample_text = "Hello World!!! This is a sample text with some 123 numbers and @special characters!   New line./n $345 "
cleaned_text = clean_text(sample_text)
print(f"Original text: {sample_text}")
print(f"Cleaned text: {cleaned_text}")
````

### Downloading NLTK Stopwords

To use NLTK's list of stopwords, you first need to download them. If you've already downloaded the entire `nltk` collection (e.g., using `nltk.download('all')`), then the stopwords are already on your system. Otherwise, you can download them specifically using the following command:

```python
nltk.download('stopwords')
```

This command fetches the default English stopwords list, among others, and makes it available for use in your NLTK environment.

````python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Ensure stopwords are downloaded (if not already done by 'all')
try:
    stopwords.words('english')
except LookupError:
    import nltk
    nltk.download('stopwords')

# Get the list of English stopwords
stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    word_tokens = word_tokenize(text)
    filtered_sentence = [w for w in word_tokens if not w.lower() in stop_words]
    return " ".join(filtered_sentence)

# Example usage with previously cleaned text
# Assuming 'cleaned_text' from the previous cell is available
sample_text_for_stopwords = cleaned_text if 'cleaned_text' in locals() else "This is an example sentence demonstrating stopword removal."

text_without_stopwords = remove_stopwords(sample_text_for_stopwords)

print(f"Original text (cleaned): {sample_text_for_stopwords}")
print(f"Text without stopwords: {text_without_stopwords}")
````

### Removing Spelling Mistakes

Now that `pyspellchecker` is installed, we can define a function to correct spelling errors in text. This function will tokenize the text into words, then use `spell.correction()` for each word, and finally rejoin the corrected words into a sentence.

````python
from spellchecker import SpellChecker
from nltk.tokenize import word_tokenize

# Initialize the SpellChecker
spell = SpellChecker()

def correct_spelling(text):
    words = word_tokenize(text)
    # Correct each word using pyspellchecker
    corrected_words = [spell.correction(word) for word in words]
    # Filter out None values that might occur if a word is not found or handled differently
    corrected_words = [word for word in corrected_words if word is not None]
    return " ".join(corrected_words)

# Example usage with text after stopword removal
# Assuming 'text_without_stopwords' from the previous cell is available
sample_text_for_spellcheck = text_without_stopwords if 'text_without_stopwords' in locals() else "Thiss is an exmple senence with speeling erors."

print(f"Original text (after stopword removal): {sample_text_for_spellcheck}")

text_corrected_spelling = correct_spelling(sample_text_for_spellcheck)

print(f"Text with corrected spelling: {text_corrected_spelling}")
````

### Removing Repeated Characters

Repeated characters (e.g., 'helllooo' to 'hello') can also be a source of noise in text data. We can use regular expressions to condense sequences of three or more identical characters into a single instance.

````python
import re

def remove_repeated_characters(text):
    # Replace three or more identical characters with two (or one, depending on desired strictness)
    # For example, 'helllooo' becomes 'hello'
    # This regex looks for any character (.), followed by itself ({2,}) three or more times.
    # The \1 refers back to the matched character, so 'aaa' becomes 'a'
    text = re.sub(r'(.)\1\1+', r'\1', text)
    return text

# Example usage with previously corrected text
# Assuming 'text_corrected_spelling' from the previous cell is available
sample_text_for_repeated_chars = text_corrected_spelling if 'text_corrected_spelling' in locals() else "Hellooo, thissss is a testttttt sentence."

print(f"Original text (after spell correction): {sample_text_for_repeated_chars}")

text_without_repeated_chars = remove_repeated_characters(sample_text_for_repeated_chars)

print(f"Text without repeated characters: {text_without_repeated_chars}")
````
